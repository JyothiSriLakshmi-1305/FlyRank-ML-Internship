```python
import pyarrow.compute as pc

date_min = None
date_max = None

for batch in parquet_file.iter_batches(
    columns=["report_date"],
    batch_size=200_000
):
    dates = batch.column("report_date")
    batch_min = pc.min(dates).as_py()
    batch_max = pc.max(dates).as_py()

    if batch_min is not None:
        date_min = batch_min if date_min is None else min(date_min, batch_min)
        date_max = batch_max if date_max is None else max(date_max, batch_max)

print("March file rows:", parquet_file.metadata.num_rows)
print("Minimum report date:", date_min)
print("Maximum report date:", date_max)
```

```python
availability_counts = {"gsc_true": 0, "ga4_true": 0, "both_true": 0}

for batch in parquet_file.iter_batches(
    columns=["gsc_data_available", "ga4_data_available"],
    batch_size=200_000
):
    gsc = batch.column("gsc_data_available")
    ga4 = batch.column("ga4_data_available")

    gsc_true = pc.equal(gsc, True)
    ga4_true = pc.equal(ga4, True)

    availability_counts["gsc_true"] += pc.sum(
        pc.fill_null(gsc_true, False)
    ).as_py()

    availability_counts["ga4_true"] += pc.sum(
        pc.fill_null(ga4_true, False)
    ).as_py()

    availability_counts["both_true"] += pc.sum(
        pc.fill_null(pc.and_(gsc_true, ga4_true), False)
    ).as_py()

print("GSC available:", availability_counts["gsc_true"])
print("GA4 available:", availability_counts["ga4_true"])
print("Both available:", availability_counts["both_true"])
```
